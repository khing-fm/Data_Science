### Understanding String Representation and Removing Display Quotes

When Python displays a list or a string in its raw form (e.g., in the output of a cell or when using `print()` without specific formatting), it often includes quotes to show that it's a string. These are not part of the string's actual content. If you want to see the content of the strings in `list_one` without these representation quotes, you can loop through the list and print each item.

````python
# Printing each item in list_one without the default string literal quotes
print("Content of list_one without string literal quotes:")
for item in list_one:
    print(item)
````

### Removing Actual Single Quote Characters from a String

If you have a string where single quotes are *part of the content* (e.g., `"'hello'"` instead of just `"hello"`) and you want to remove these specific quote characters, you can use the `replace()` string method or `strip()` if they are only at the beginning or end.

````python
# Example: A string that literally contains single quotes
string_with_quotes = "'This is a quoted string.'"
print(f"Original string: {string_with_quotes}")

# Using .replace() to remove all single quote characters
string_without_quotes_replace = string_with_quotes.replace("'", "")
print(f"Using replace(): {string_without_quotes_replace}")

# Example: A string with leading/trailing quotes
string_with_outer_quotes = "' another example '"
print(f"Original string with outer quotes: {string_with_outer_quotes}")

# Using .strip() to remove leading/trailing single quote characters (and spaces by default)
string_without_outer_quotes_strip = string_with_outer_quotes.strip(" '") # strip both spaces and single quotes
print(f"Using strip(): {string_without_outer_quotes_strip}")
````

### List Operators

In Python, you can use various operators with lists to perform common operations like concatenation, repetition, membership testing, and slicing.

1.  **Concatenation (`+`)**: Combines two or more lists into a single new list.
2.  **Repetition (`*`)**: Repeats a list a specified number of times to create a new list.
3.  **Membership (`in`, `not in`)**: Checks if an item exists within a list.
4.  **Slicing (`:`)**: Extracts a portion (a 'slice') of a list.
5.  **Equality (`==`, `!=`)**: Compares two lists to see if they are equal (have the same elements in the same order) or not equal.

Let's see some examples:

````python
# Example lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = ['a', 'b', 'c']

print("Original lists:")
print(f"list1: {list1}")
print(f"list2: {list2}")
print(f"list3: {list3}")
print("\n-- List Operators in Action --")

# 1. Concatenation (+)
combined_list = list1 + list2
print(f"\n1. Concatenation (list1 + list2): {combined_list}")

# 2. Repetition (*)
repeated_list = list3 * 2
print(f"2. Repetition (list3 * 2): {repeated_list}")

# 3. Membership (in, not in)
is_two_in_list1 = 2 in list1
is_seven_not_in_list1 = 7 not in list1
print(f"\n3. Membership (2 in list1): {is_two_in_list1}")
print(f"   Membership (7 not in list1): {is_seven_not_in_list1}")

# 4. Slicing (:)
# Syntax: list[start:end:step]
first_two_elements = list1[0:2] # or list1[:2]
last_element = list2[-1]
middle_elements = combined_list[2:5]
print(f"\n4. Slicing (list1[0:2]): {first_two_elements}")
print(f"   Slicing (list2[-1]): {last_element}")
print(f"   Slicing (combined_list[2:5]): {middle_elements}")

# 5. Equality (==, !=)
are_lists_equal = list1 == [1, 2, 3]
are_lists_not_equal = list1 != list2
print(f"\n5. Equality (list1 == [1, 2, 3]): {are_lists_equal}")
print(f"   Equality (list1 != list2): {are_lists_not_equal}")
````

### `min()` and `max()` Functions

Python's built-in `min()` and `max()` functions are used to find the smallest and largest item in an iterable (like a list, tuple, or string) or among multiple arguments.

-   **`min(iterable)`**: Returns the smallest item in an iterable.
-   **`max(iterable)`**: Returns the largest item in an iterable.

They can also take multiple arguments directly:

-   **`min(arg1, arg2, ...)`**: Returns the smallest of the given arguments.
-   **`max(arg1, arg2, ...)`**: Returns the largest of the given arguments.

Let's look at some examples.

````python
# Example with a list of numbers
numbers = [10, 4, 25, 7, 13]
print(f"List of numbers: {numbers}")
print(f"Minimum number: {min(numbers)}")
print(f"Maximum number: {max(numbers)}")

# Example with a list of strings (lexicographical comparison)
words = ["apple", "zebra", "banana", "cat"]
print(f"\nList of words: {words}")
print(f"Minimum word: {min(words)}")
print(f"Maximum word: {max(words)}")

# Example with multiple arguments
print(f"\nMinimum among (5, 12, 3, 8): {min(5, 12, 3, 8)}")
print(f"Maximum among (5, 12, 3, 8): {max(5, 12, 3, 8)}")

# Example with an empty list (will raise an error)
# empty_list = []
# try:
#     min(empty_list)
# except ValueError as e:
#     print(f"\nError for empty list: {e}")

# Example with a custom key function (e.g., find the longest word)
names = ["Alice", "Bob", "Charlie", "David"]
print(f"\nList of names: {names}")
longest_name = max(names, key=len)
shortest_name = min(names, key=len)
print(f"Longest name (using key=len): {longest_name}")
print(f"Shortest name (using key=len): {shortest_name}")
````

### `sum()`, `sort()` Method, and `sorted()` Function

These are essential tools for performing common operations on lists and other iterables in Python.

#### 1. `sum()` Function

-   **Purpose**: The `sum()` function is used to calculate the sum of all numerical items in an iterable (like a list, tuple, or set).
-   **Syntax**: `sum(iterable, start=0)`
    -   `iterable`: The collection of numbers to sum.
    -   `start` (optional): An initial value to which the items of the iterable are added. Defaults to 0.
-   **Returns**: The total sum as a number.

#### 2. `list.sort()` Method

-   **Purpose**: The `sort()` method is a list method that sorts the items of the list **in-place**. This means it modifies the original list directly.
-   **Syntax**: `list.sort(key=None, reverse=False)`
    -   `key` (optional): A function to be called on each list item prior to making comparisons. (e.g., `key=len` for sorting by length of strings).
    -   `reverse` (optional): If `True`, the list elements are sorted in descending order. Defaults to `False` (ascending order).
-   **Returns**: `None` (because it modifies the list in-place).

#### 3. `sorted()` Function

-   **Purpose**: The `sorted()` function is a built-in function that sorts the items of an iterable and returns a **new sorted list**. It does not modify the original iterable.
-   **Syntax**: `sorted(iterable, key=None, reverse=False)`
    -   `iterable`: The collection of items to be sorted.
    -   `key` (optional): Same as for `list.sort()`.
    -   `reverse` (optional): Same as for `list.sort()`.
-   **Returns**: A new list containing all items from the iterable in sorted order.

Let's look at examples for each:

````python
# --- 1. sum() Function ---
my_numbers = [1, 5, 2, 8, 3]

print(f"Original numbers list: {my_numbers}")

# Summing all elements
total_sum = sum(my_numbers)
print(f"Sum of numbers: {total_sum}")

# Summing with a start value
total_sum_with_start = sum(my_numbers, 10)
print(f"Sum of numbers with start=10: {total_sum_with_start}")


# --- 2. list.sort() Method (in-place) ---
my_list_to_sort = [3, 1, 4, 1, 5, 9, 2, 6]
my_words = ["banana", "apple", "kiwi", "grape"]

print(f"\nOriginal list for sort(): {my_list_to_sort}")
my_list_to_sort.sort() # Sorts in-place (ascending)
print(f"List after my_list_to_sort.sort(): {my_list_to_sort}")

my_list_to_sort_desc = [3, 1, 4, 1, 5, 9, 2, 6]
my_list_to_sort_desc.sort(reverse=True) # Sorts in-place (descending)
print(f"List after my_list_to_sort_desc.sort(reverse=True): {my_list_to_sort_desc}")

print(f"Original words list for sort(): {my_words}")
my_words.sort(key=len) # Sorts by length of words
print(f"Words after my_words.sort(key=len): {my_words}")


# --- 3. sorted() Function (returns new list) ---
another_numbers_list = [9, 8, 7, 6, 5]
more_words = ["strawberry", "apple", "fig", "blueberry"]

print(f"\nOriginal list for sorted(): {another_numbers_list}")
sorted_numbers = sorted(another_numbers_list) # Returns new sorted list
print(f"New list after sorted(another_numbers_list): {sorted_numbers}")
print(f"Original list remains unchanged: {another_numbers_list}")

sorted_numbers_desc = sorted(another_numbers_list, reverse=True)
print(f"New list after sorted(..., reverse=True): {sorted_numbers_desc}")

print(f"Original words list for sorted(): {more_words}")
sorted_words_by_length = sorted(more_words, key=len)
print(f"New words list after sorted(..., key=len): {sorted_words_by_length}")
````

````python
# The `range()` function
# `range()` generates a sequence of numbers, which is commonly used in for loops.
# It does not store all the numbers in memory, making it efficient for large sequences.

# Syntax: range(stop), range(start, stop), range(start, stop, step)

# Example 1: range(stop) - starts from 0 up to (but not including) stop
print("Range from 0 to 5 (exclusive):")
for i in range(5):
    print(i, end=" ")
print("\n")

# Example 2: range(start, stop) - starts from start up to (but not including) stop
print("Range from 2 to 7 (exclusive):")
for i in range(2, 7):
    print(i, end=" ")
print("\n")

# Example 3: range(start, stop, step) - starts from start, increments by step, up to (but not including) stop
print("Range from 0 to 10 with step 2:")
for i in range(0, 10, 2):
    print(i, end=" ")
print("\n")

print("Range from 10 to 0 with step -1 (countdown):")
for i in range(10, 0, -1):
    print(i, end=" ")
print("\n")

# Converting a range object to a list to see its contents
print(f"List from range(5): {list(range(5))}")
print(f"List from range(1, 6): {list(range(1, 6))}")
print(f"List from range(0, 10, 3): {list(range(0, 10, 3))}")

# --- Slicing revisited, potentially with range for context ---
# Slicing allows you to extract a portion (a 'slice') of a sequence (like lists, strings, tuples).
# Syntax: sequence[start:end:step]

my_list = list(range(10, 101, 10)) # Create a list using range for demonstration
print(f"\nOriginal list created with range(): {my_list}")

# Basic slicing
print(f"Slice elements from index 2 to 5 (exclusive): {my_list[2:6]}") # [30, 40, 50, 60]

# Slicing from the beginning
print(f"Slice elements from beginning to index 3 (exclusive): {my_list[:3]}") # [10, 20, 30]

# Slicing to the end
print(f"Slice elements from index 5 to the end: {my_list[5:]}") # [60, 70, 80, 90, 100]

# Slicing with a step
print(f"Slice every other element: {my_list[::2]}") # [10, 30, 50, 70, 90]

# Reverse the list using slicing
print(f"Reversed list using slicing: {my_list[::-1]}") # [100, 90, 80, 70, 60, 50, 40, 30, 20, 10]

# Using range to iterate over slices or generate indices for operations
# Example: Process elements at even indices
even_indexed_elements = []
for i in range(0, len(my_list), 2):
    even_indexed_elements.append(my_list[i])
print(f"Elements at even indices (using range to generate indices): {even_indexed_elements}")

# This demonstrates how range generates sequences (often for indices)
# and slicing uses those index-like concepts to extract parts of sequences.
````

````python
### Tuples

"""Tuples are ordered, immutable collections of items. This means:

*   **Ordered**: The items have a defined order, and that order will not change.
*   **Immutable**: Once a tuple is created, you cannot change its elements (add, remove, or modify).
*   **Allow Duplicates**: Tuples can contain duplicate values.
*   **Indexed**: You can access tuple items by referring to their index number (starting from 0).

Tuples are often used for heterogeneous data types (different types of data) and are typically created with parentheses `()`.

```python"""
# Creating a tuple
my_tuple = (1, "hello", 3.14, True, 1)
print(f"My tuple: {my_tuple}")
print(f"Type of my_tuple: {type(my_tuple)}")

# Accessing elements
print(f"First element: {my_tuple[0]}")
print(f"Second element: {my_tuple[1]}")

# Tuples allow duplicates
print(f"Tuple with duplicates: {my_tuple}")

# Attempting to change an element (will raise an error)
# try:
#     my_tuple[0] = 5
# except TypeError as e:
#     print(f"Error when modifying tuple: {e}")

# Tuple concatenation
tuple1 = (1, 2)
tuple2 = (3, 4)
combined_tuple = tuple1 + tuple2
print(f"Combined tuple: {combined_tuple}")

# Checking for membership
print(f"'hello' in my_tuple: {'hello' in my_tuple}")

### Sets

"""Sets are unordered, mutable collections of unique items. This means:

*   **Unordered**: The items in a set do not have a defined order, and their order can change.
*   **Mutable**: You can add and remove items from a set after it has been created.
*   **No Duplicates**: Sets automatically remove duplicate values. Each element in a set must be unique.
*   **Not Indexed**: Because sets are unordered, you cannot access items by referring to an index or key.

Sets are typically created with curly braces `{}` or using the `set()` constructor. An empty set must be created using `set()`, as `{}` creates an empty dictionary."""

#```python
# Creating a set
my_set = {1, "apple", 3.14, True, 1, "apple"} # Duplicates 1 and "apple" will be removed
print(f"My set: {my_set}")
print(f"Type of my_set: {type(my_set)}")

# Creating an empty set
empty_set = set()
print(f"Empty set: {empty_set}")

# Adding elements to a set
my_set.add(5)
print(f"Set after adding 5: {my_set}")
my_set.add("orange")
print(f"Set after adding 'orange': {my_set}")

# Adding an existing element (no change occurs)
my_set.add(1)
print(f"Set after adding 1 again: {my_set}")

# Removing elements from a set
my_set.remove(1)
print(f"Set after removing 1: {my_set}")

# Checking for membership
print(f"'apple' in my_set: {'apple' in my_set}")

# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(f"\nSet A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union (A | B): {set_a | set_b}") # All unique elements from both sets
print(f"Intersection (A & B): {set_a & set_b}") # Common elements
print(f"Difference (A - B): {set_a - set_b}") # Elements in A but not in B
print(f"Symmetric Difference (A ^ B): {set_a ^ set_b}") # Elements in A or B but not both
````

````python
def calculate_payout_and_fee(amount_str):
  try:
    amount = float(amount_str)
    rate = 350 / 10000
    fee = amount * rate
    payout = amount - fee
    return fee, payout # Return the calculated fee and payout
  except ValueError:
    print("Invalid input. Please enter a numerical amount.")
    return None, None # Return None for both if there's an error

# Now, let's use the function
user_input = input("Enter Amount")
calculated_fee, calculated_payout = calculate_payout_and_fee(user_input)

if calculated_fee is not None and calculated_payout is not None:
  print(f"Fee: {calculated_fee:.0f}, Payout: {calculated_payout:.0f}")
````
