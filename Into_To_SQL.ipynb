## Persists a database to avoid dataset getting lost after closing

```python
path = "demo.db"
if os.path.exists(path):
    os.remove(path)

conn_c = sqlite3.connect(path)
conn_c.execute("CREATE TABLE t (x INTEGER)")
conn_c.execute("INSERT INTO t (x) VALUES (1)")
conn_c.commit()
conn_c.close()

conn_d = sqlite3.connect(path)   # a new connection, same file
print(pd.read_sql_query("SELECT * FROM t", conn_d))
print("file really exists on disk:", os.path.exists(path), "_", os.path.getsize(path), "bytes")

conn_d.close()
os.remove(path)  # Cleanup, this was just a demonstration of how to create a file on disk and read it back in.
```
