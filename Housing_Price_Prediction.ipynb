## 📏 Model Evaluation

After training the models, their performance will be evaluated using several regression metrics.

The main evaluation metrics will be:

### Mean Absolute Error (MAE)

Mean Absolute Error measures the average absolute difference between the actual house values and the values predicted by the model.

A lower MAE indicates better performance.

The MAE can be interpreted as the approximate average amount by which the model's predictions differ from the actual house values.

---

### Root Mean Squared Error (RMSE)

Root Mean Squared Error measures the square root of the average squared prediction errors.

RMSE gives greater importance to larger prediction errors than MAE.

A lower RMSE indicates better model performance.

---

### R² Score

The R² score measures the proportion of variation in the target variable that can be explained by the model.

A higher R² score generally indicates better predictive performance.

The three metrics will be considered together when comparing the models rather than relying on a single metric.

`````python
# ==========================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ==========================================

# Data manipulation and numerical operations
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine learning: train/test split
from sklearn.model_selection import train_test_split

# Data preprocessing
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

# Machine learning models
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Model evaluation
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("All libraries imported successfully!")
`````

`````python
# Load dataset

path = r"C:\Users\EHU_HEALTH_TECHNICAL\Documents\Khing_FM_Data_Science\Data_Science\assets\ts-academy_datasets\housing.csv"

df = pd.read_csv(path)
`````

`````python
# Let's Define X and Y

# Features
x = df.drop(
    'median_house_value',
    axis=1
)

#Target
y = df['median_house_value']

print("x shape:", x.shape)
print("y shape:", y.shape)
`````

`````python
# Identifying numerical and categorical variables

numerical_features = [
    'longitude',
    'latitude',
    'housing_median_age',
    'total_rooms',
    'total_bedrooms',
    'population',
    'households',
    'median_income',
]

categorical_features = [
    'ocean_proximity'
]

print("Numerical features:", numerical_features)
print("\nCategorical features:", categorical_features)
`````

`````python
# Training and Testing
# Split data first

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", x_train.shape)
print("Testing features:", x_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
`````

`````python
# Data processing
# Numerical processing (Use pipeline processing)
# For missing numerical values, replace them with the median of that column.

numerical_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median'))
    ]
)
`````

`````python
# Create the categorical transformer

categorical_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('oneshot', OneHotEncoder(handle_unknown='ignore'))
    ]
)
`````

`````python
# Combine the processing

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
`````

`````python
# Now, let's create a model

linear_model = LinearRegression()
`````

`````python
# Complete ML pipeline

linear_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', linear_model)
    ]
)
`````

`````python
# Train Model

linear_pipeline.fit(
    x_train,
    y_train
)
`````

`````python
# Making predictions

y_pred_linear = linear_pipeline.predict(x_test)

print(y_pred_linear[:10])
`````

`````python
# Model Evaluation
# Mean Absolute Error (MAE)

mae_linear = mean_absolute_error(
    y_test,
    y_pred_linear
)

print(f"Linear Regression MAE: ${mae_linear:,.2f}")
`````

`````python
# Relative Mean Squared Error (RMNSE)

rmnse_linear = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_linear
    )
)

print(f"Linear Regression RMNSE: ${rmnse_linear:,.2f}")
`````

`````python
# R-Squared (R²)

r2_linear = r2_score(
    y_test,
    y_pred_linear
)

print(f"Linear Regression R²: {r2_linear:.4f}")
`````

| Metric   |    Result | What it means                                                        |

| **RMSE** | **$70,059.19** | Larger prediction errors are penalized more heavily                  |

`````python
# Decision Tree Regression

decision_tree_model = DecisionTreeRegressor(
    random_state=42
)

decision_tree_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', decision_tree_model)
    ]
)

# Train the model
decision_tree_pipeline.fit(
    x_train,
    y_train
)
`````

`````python
# Predictions

y_pred_tree = decision_tree_pipeline.predict(x_test)

print(y_pred_tree[:10])
`````

`````python
# Evaluate MAE, RMSE, and R²

# MAE
mae_tree = mean_absolute_error(
    y_test,
    y_pred_tree
)

print(f"Decision Tree MAE: ${mae_tree:,.2f}")

# RMSE
rmse_tree = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_tree
    )
)

print(f"Decision Tree RMSE: ${rmse_tree:,.2f}")

# R²
r2_tree = r2_score(
    y_test,
    y_pred_tree
)

print(f"Decision Tree R²: ${r2_tree:,.2f}")
`````

`````python
# Random forest regression

random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
`````

`````python
# Random Forest pipeline
random_forest_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', random_forest_model)
    ]
)
`````

`````python
# Train the Random Forest

random_forest_pipeline.fit(
    x_train,
    y_train
)
`````
