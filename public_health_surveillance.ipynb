```python
# Gender distribution chart
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Gender"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Individuals")

plt.show()
```

```python
# Vaccination analysis and Visualization
# Zero for Non Vaccinated and One for Vaccinated

print(
    df["Vaccination_Status"].value_counts()
)

print(
    df["Vaccination_Status"]
    .value_counts(normalize=True)
    * 100
)

# Vaccination Status Distribution Chart
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Vaccination_Status"
)

plt.title("Vaccination Status")
plt.xlabel("Vaccination Status")
plt.ylabel("Number of Individuals")

plt.show()
```

```python
# Vaccination vs Infection Risk

vaccination_risk = pd.crosstab(
    df["Vaccination_Status"],
    df["Infection_Risk_Level"]
)

print(vaccination_risk)

#Percentage of Infection Risk by Vaccination Status
vaccination_risk_percentage = pd.crosstab(
    df["Vaccination_Status"],
    df["Infection_Risk_Level"],
    normalize="index"
) * 100

print(vaccination_risk_percentage)

#Visualization of Vaccination vs Infection Risk
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Vaccination_Status",
    hue="Infection_Risk_Level"
)

plt.title("Vaccination Status and Infection Risk")
plt.xlabel("Vaccination Status")
plt.ylabel("Number of Individuals")

plt.show()
```

```python
# Disease Severity
severity = df["Disease_Severity"].value_counts()

print(severity)

# Plot
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Disease_Severity"
)

plt.title("Disease Severity Distribution")
plt.xlabel("Disease Severity")
plt.ylabel("Number of Individuals")

plt.show()
```

```python
# Chronic conditions vs disease severity
chronic_severity = pd.crosstab(
    df["Chronic_Conditions"],
    df["Disease_Severity"]
)

print(chronic_severity)

#Visualization of Chronic Conditions vs Disease Severity
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Chronic_Conditions",
    hue="Disease_Severity"
)

plt.title("Chronic Conditions vs Disease Severity")
plt.xlabel("Chronic Conditions")
plt.ylabel("Number of Individuals")

plt.show()
```

```python
# outbreak status analysis

outbreak_status = df["Outbreak_Status"].value_counts()

print(outbreak_status)

# Plot
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Outbreak_Status"
)

plt.title("Outbreak Status Distribution")
plt.xlabel("Outbreak Status")
plt.ylabel("Number of Individuals")

plt.show()
```
