```python
# Example: Creating a Series

# From a list
sales = pd.Series([100, 150, 200, 175, 225])
print("series from list:")
print(sales)
print(sales.index)
print(f"\nIndex: {sales.index.tolist()}")
print(f"Values: {sales.values.tolist()}")
```

```python
sales_mean = sales.mean()
sales_std = sales.std()

print(f"Mean Sales: {sales_mean}")
print(f"Standard Deviation of Sales: {sales_std}")
```

### Sales Data Analysis Summary

*   **Mean Sales (170.0)**: Represents the average value of the sales entries.
*   **Standard Deviation (~48.09)**: Indicates the spread of the data; a higher value suggests that the sales figures vary significantly from the average.

### Creating a Series with NumPy and dtypes

This example demonstrates:
1.  **NumPy Integration**: Creating a Series directly from an `np.array`.
2.  **Memory Optimization**: Using `dtype="int16"` to store integers using less memory (16 bits) compared to the standard 64-bit default.

```python
# Example: Series with custom index

sales = pd.Series(
    [100, 150, 200, 175, 225],
    index=["Jan", "Feb", "Mar", "Apr", "May"]
)
print("series with custom index:")
print(sales)
print(sales.index)
print(f"\nIndex: {sales.index.tolist()}")
print(f"Values: {sales.values.tolist()}")

# Acess by label
print(f"\nSales in May: {sales['May']}")
print(f"Jan to Mar: \n{sales['Jan':'Mar']}")
```

### Custom Indexing and Label-Based Access

Pandas allows you to define a **custom index** (labels) for your data instead of just using default integers:

*   **Custom Labels**: In this example, we used month names (`Jan`, `Feb`, etc.) as the index.
*   **Access by Label**: You can retrieve specific values using the label name, like `sales['May']`.
*   **Slicing with Labels**: Unlike Python lists, when you slice with labels (`'Jan':'Mar'`), the **end point is included** in the result.
