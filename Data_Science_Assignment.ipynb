```python
# ==========================================================
# Pandas Data Analysis Assessment
# ==========================================================

# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Display plots inside Jupyter Notebook
%matplotlib inline

# ==========================================================
# Section A – Basic Exploration
# ==========================================================

# 1. Read the StudentsPerformance.csv dataset
df = pd.read_csv("/content/sample_data/StudentsPerformance.csv")

# 2. Display the first 10 rows
print("First 10 Rows")
display(df.head(10))

# Display the last 10 rows
print("Last 10 Rows")
display(df.tail(10))

# 3. Check dataset overview
print("\nDataset Information")
df.info()

# 4. Check dataset shape
print("\nDataset Shape:")
print(df.shape)

# 5. List all column names
print("\nColumn Names:")
print(df.columns.tolist())

# 6. Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(exclude=np.number).columns

print("\nNumerical Columns:")
print(numerical_columns.tolist())

print("\nCategorical Columns:")
print(categorical_columns.tolist())

# 7. Descriptive statistics
print("\nDescriptive Statistics")
display(df.describe())

# 8. Check for missing values
print("\nMissing Values")
print(df.isnull().sum())
```

```python
# ==========================================================
# Section B – Filtering & Conditional Selection
# ==========================================================

# 1. Students who completed the test preparation course
completed = df[df['test preparation course'] == 'completed']

print("\nStudents who completed test preparation")
display(completed)

# 2. Students scoring above 70 in math
math_above70 = df[df['math score'] > 70]

print("\nStudents with Math Score Above 70")
display(math_above70)

# 3. Female students with at least 65 in all three subjects
female_65 = df[
    (df['gender'] == 'female') &
    (df['math score'] >= 65) &
    (df['reading score'] >= 65) &
    (df['writing score'] >= 65)
]

print("\nFemale Students Scoring At Least 65 in All Subjects")
display(female_65)

# 4. Count students with free/reduced lunch
free_lunch = df[df['lunch'] == 'free/reduced']

print("\nNumber of Students with Free/Reduced Lunch:")
print(free_lunch.shape[0])
```

```python
# ==========================================================
# Section C – GroupBy & Aggregation
# ==========================================================

# 1. Average math score by gender
avg_math_gender = df.groupby('gender')['math score'].mean()

print("\nAverage Math Score by Gender")
print(avg_math_gender)

# 2. Average scores by parental education
avg_parent = df.groupby('parental level of education')[
    ['math score', 'reading score', 'writing score']
].mean()

print("\nAverage Scores by Parental Level of Education")
display(avg_parent)

# 3. Average scores by test preparation
avg_testprep = df.groupby('test preparation course')[
    ['math score', 'reading score', 'writing score']
].mean()

print("\nAverage Scores by Test Preparation")
display(avg_testprep)
```

```python
# ==========================================================
# Section D – Visualization
# ==========================================================

# Histogram - Math Score
plt.figure(figsize=(6,4))
plt.hist(df['math score'], bins=10)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.show()

# Histogram - Reading Score
plt.figure(figsize=(6,4))
plt.hist(df['reading score'], bins=10)
plt.title('Distribution of Reading Scores')
plt.xlabel('Reading Score')
plt.ylabel('Frequency')
plt.show()

# Histogram - Writing Score
plt.figure(figsize=(6,4))
plt.hist(df['writing score'], bins=10)
plt.title('Distribution of Writing Scores')
plt.xlabel('Writing Score')
plt.ylabel('Frequency')
plt.show()

# Average score by gender
avg_math_gender.plot(kind='bar', figsize=(6,4))

plt.title('Average Math Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Math Score')
plt.xticks(rotation=0)
plt.show()
```

```python
# ==========================================================
# Section E – Interpretation
# ==========================================================

print("""
Interpretation:

Students who completed the test preparation course generally have
higher average scores in mathematics, reading, and writing compared
to students who did not complete the course. This suggests that
participating in the test preparation course positively influences
academic performance. However, other factors such as parental
education, study habits, and socioeconomic background may also
contribute to students' performance.
""")
```
